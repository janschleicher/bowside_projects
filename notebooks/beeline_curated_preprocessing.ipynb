```python
from pathlib import Path

import networkx as nx
import numpy as np
import palantir
import pandas as pd
import scanpy as sc
import seaborn as sns
from matplotlib import pyplot as plt
```

```python
# Load data
names = ["mCAD", "VSC", "HSC", "GSD"]
dropout = 0  # options: 0, 50, 70
adatas = {}
adj_lists = {}
for name in names:
    input_dir = f"../data/BEELINE-data/inputs/Curated/{name}{'-q' + str(dropout) if dropout > 0 else ''}"
    adatas[name] = []
    for i in range(10):
        data_dir = f"{input_dir}/{name}-2000-{i+1}{'-' + str(dropout) if dropout > 0 else ''}/"
        exp_df = pd.read_csv(f"{data_dir}/ExpressionData.csv", index_col=0).T
        obs_df = pd.read_csv(f"{data_dir}/PseudoTime.csv", index_col=0)
        # Only one of the n pseudotimes is not NA
        assert np.all((~obs_df.isna()).sum(axis=1) == 1)
        obs_df["pseudotime"] = obs_df.fillna(0).sum(axis=1)
        adata = sc.AnnData(X=exp_df, obs=obs_df)
        display(adata)
        adatas[name].append(adata)
    adj_list = pd.read_csv(f"{input_dir}/GroundTruthNetwork.csv")
    display(adj_list)
    adj_lists[name] = adj_list
```

```python
# Preprocessing
for lst in adatas.values():
    for adata in lst:
        adata.layers["log1p"] = adata.X.copy()
        sc.pp.scale(adata)
        sc.pp.neighbors(adata, n_neighbors=30)
        sc.tl.umap(adata)
```

```python
# MAGIC imputation
if dropout > 0:
    for lst in adatas.values():
        for adata in lst:
            _ = palantir.utils.run_diffusion_maps(adata, n_components=5, knn=30)
            _ = palantir.utils.determine_multiscale_space(adata)
            _ = palantir.utils.run_magic_imputation(adata, expression_key="log1p")
```

```python
fig, axs = plt.subplots(
    len(names),
    len(adatas[names[0]]),
    figsize=(len(adatas[names[0]]) * 2, len(names) * 2),
    layout="constrained",
)
for i, (name, lst) in enumerate(adatas.items()):
    for j, adata in enumerate(lst):
        sc.pl.umap(
            adata,
            color="pseudotime",
            cmap="viridis",
            colorbar_loc=None,
            s=5,
            title=f"{name} {j+1}",
            ax=axs[i, j],
            show=False,
        )
        axs[i, j]
```

```python
for name, lst in adatas.items():
    adata = lst[0]
    ncol = 4
    nrow = int(np.ceil(adata.shape[1] / ncol))
    fig, axs = plt.subplots(
        nrow, ncol, figsize=(ncol * 3, nrow * 2.5), layout="constrained"
    )
    for i, gene in enumerate(adata.var_names):
        sc.pl.scatter(
            adata,
            x="pseudotime",
            y=gene,
            layers="MAGIC_imputed_data" if dropout > 0 else "log1p",
            ax=axs.flat[i],
            size=20,
            show=False,
            title="",
            use_raw=False,
            alpha=0.1,
        )
    for ax in axs.flat[adata.shape[1] :]:
        ax.set_axis_off()

    fig.suptitle(name)
    sns.despine(fig)
```

```python
# Make input GRN
grns = {}
for name, adj_list in adj_lists.items():
    grns[name] = nx.from_pandas_edgelist(
        adj_list,
        source="Gene1",
        target="Gene2",
        edge_attr="Type",
        create_using=nx.DiGraph,
    )
```

```python
fig, axs = plt.subplots(
    1, len(names), figsize=(4 * len(names), 4), layout="constrained"
)
for ax, name in zip(axs, names):
    nx.draw_circular(grns[name], with_labels=True, ax=ax)
```

```python
out_dir_overall = Path("../data/BEELINE-data/processed/Curated")
out_dir_overall.mkdir(parents=True, exist_ok=True)
```

```python
for name, lst in adatas.items():
    out_dir = out_dir_overall / name
    out_dir.mkdir(parents=True, exist_ok=True)
    adj_lists[name].to_csv(out_dir / f"{name}_adj_list.csv")
    for i, adata in enumerate(lst):
        adata.write_h5ad(out_dir / f"{name}_processed_{i+1}.h5ad", compression="gzip")
```

```python
for lst in adatas.values():
    for adata in lst:
        assert np.all(sorted(adata.var_names) == adata.var_names.to_list())
```
